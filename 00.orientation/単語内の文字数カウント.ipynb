```python
listfirst=[]
dictionary= dict.fromkeys(list("abcdefghijklmnopqrstuvwxyz"), 0)
while True:
    a=input(f"英単語を入力してください:")
    listfirst.append(a)
    listfirst.sort()
    if a=="":
        break
    print(f"入力した英単語：{listfirst}")
    for b in a:
        if b in dictionary:
            dictionary[b] += 1
for b in dictionary:
        if dictionary[b]>=1:
             print(f"{b}が{dictionary[b]}個ありました。")
```
