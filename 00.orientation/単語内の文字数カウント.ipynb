```python
listfirst=[]
dictionary= dict.fromkeys(list("abcdefghijklmnopqrstuvwxyz"), 0)
while True:
    a=input(f"英単語を入力してください:{a}")
    listfirst.append(a)
    listfirst.sort()
    print(f"入力した英単語：{listfirst}")
    if a=="":
        break
    for char in a:
        if char in dictionary:
            dictionary[char] += 1
for char in dictionary:
        if dictionary[char]>=1:
             print(f"{char}が{dictionary[char]}個ありました。")
```
