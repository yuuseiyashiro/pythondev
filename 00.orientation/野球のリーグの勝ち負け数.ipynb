```python
print("team     win lose draw rate")
class Baseballteam:
    name=""
    win=0
    Lose=0
    Draw=0
    def __init__(self,name,win,lose,draw):
        self.name=name
        self.win=win
        self.Lose=lose
        self.Draw=draw
def calc_win_rate():
    team1,team2,team3,team4,team5,team6=0,0,0,0,0,0
    team1=Baseballteam("Archers",77,64,2)
    team2=Baseballteam("Meteors",71,69,3)
    team3=Baseballteam("Cheetahs",69,68,6)
    team4=Baseballteam("Sharks",70,70,3)
    team5=Baseballteam("Pythons",68,73,2)
    team6=Baseballteam("Owls",59,82,2)
def show_team_result():
    teams=[team1,team2,team3,team4,team5,team6]
    for team in teams:
        win_rate=team.win/(team.win+team.Lose)
        print(f"{team.name}    {team.win}   {team.Lose}   {team.Draw} {win_rate:.3f}")
calc_win_rate()
show_team_result()
```
